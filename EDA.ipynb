```python
from pathlib import Path

TREES_PATH = Path("data/ofo_ground-reference_trees.gpkg")
if TREES_PATH.stat().st_size == 0:
    raise FileNotFoundError(
        f"{TREES_PATH} is empty. Download the file from "
        "https://zenodo.org/records/22731179/files/ofo_ground-reference_trees.gpkg?download=1"
    )

gpd.list_layers(TREES_PATH)
trees = gpd.read_file(TREES_PATH)
trees.head()
trees.crs
trees.geom_type.unique()
trees.columns
```
